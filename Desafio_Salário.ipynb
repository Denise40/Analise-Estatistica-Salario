```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# lendo o arquivo csv
df = pd.read_csv('/content/salario_profissionais_dados.csv.csv')

# mostrando as 5 primeiras linhas do data frame
df.head()
```

```python
# verificando se há dados nulos e o tipo de dados de cada coluna
df.info()
```

```python
# Verificando número de linhas e colunas
df.shape
```

```python
# verificando as estatísticas descritivas básicas do dataset de forma geral
df.describe()
```

```python
#verificando os valores nulos
df.isnull().sum()
```

```python
# cargos mais frequentes (Top 10)
print('Cargos mais comuns:')
print(df['job_title'].value_counts().head(10))
```

```python
# proporção relativa aos cargos mais comuns
print("\nProporção relativa dos cargos:")
print(df['job_title'].value_counts(normalize=True).head(10) * 100)
```

```python
#distribuição em gráfico de cargos mais comuns
df['job_title'].value_counts().head(10).plot(kind='bar', color='skyblue')
plt.show()
```

```python
# nível de experiência
print('Nível de experiência:')
print(df['experience_level'].value_counts())
```

```python
# proporção relativa ao nível de experiência
print("\nProporção dos níveis de experiência:")
print(df['experience_level'].value_counts(normalize=True) * 100)
```

```python
#distribuição em gráfico do nível de experiência
df['experience_level'].value_counts().plot(kind='bar', color = 'skyblue')
plt.show()
```

```python
#tamanho das empresas
print('Tamanho da empresa:')
print(df['company_size'].value_counts())
```

```python
# proporção relativa ao tamanho das empresas
print("\nProporção do tamanho da empresa:")
print(df['company_size'].value_counts(normalize=True) * 100)
```

```python
#distribuição em gráfico do tamanho das empresas
df['company_size'].value_counts().plot(kind='bar', color='skyblue')
plt.show()
```

```python
#calculando a média do salário anual em dólares
media = df['salary_in_usd'].mean()
print(f"Média salarial: ${media:.2f}")
```

```python
#calculando a mediana do salário anual em dólares
mediana = df['salary_in_usd'].median()
print(f"Mediana salarial: ${mediana:.2f}")
```

```python
#calculando o desvio padrão do salário anual em dólares
desvio_padrao = df['salary_in_usd'].std()
print(f"Desvio padrão: ${desvio_padrao:.2f}")
```

```python
#valor máximo do salário anual
maximo = df['salary_in_usd'].max()
print(f"Valor máximo: ${maximo:.2f}")
```

```python
#valor mínimo do salário anual
minimo = df['salary_in_usd'].min()
print(f"Valor mínimo: ${minimo:.2f}")
```

```python
# hsitograma - distribuição geral
plt.figure(figsize=(10, 6))
sns.histplot(df['salary_in_usd'], bins=30, kde=True, color='skyblue')
plt.title('Distribuição Salarial (USD)')
plt.xlabel('Salário Anual em USD')
plt.ylabel('Frequência')
plt.show()
```

```python
# boxplot dos salários por nível de experiência
plt.figure(figsize=(10, 6))
sns.boxplot(x='experience_level', y='salary_in_usd', data=df, color='Skyblue')
plt.title('Distribuição de Salários por Nível de Experiência')
plt.xlabel('Nível de Experiência')
plt.ylabel('Salário em USD')
plt.show()
```

```python
# comparar média salarial por nível de experiência
media_por_experiencia = df.groupby('experience_level')['salary_in_usd'].mean().round(2).sort_values(ascending=False)
print("Média salarial por nível de experiência:")
print(media_por_experiencia)
```

```python
# Calcular a média salarial por país (Top 10)
media_por_pais = df.groupby('country')['salary_in_usd'].mean().round(2).sort_values(ascending=False).head(10)
print("Top 10 países com maiores médias salariais:")
print(media_por_pais)
```

```python
# análise de correlação com matriz

# Selecionar colunas quantitativas relevantes
colunas_numericas = df[['salary_in_usd', 'work_year', 'years_of_experience']]

# Calcular matriz de correlação
matriz_correlacao = colunas_numericas.corr().round(2)

#Exibir matriz
print("Matriz de correlação:")
print(matriz_correlacao)
```

Correlação salary_in_usd vs. years_of_experience:

Correlação: 0.41

Pode-se perceber que há uma correlação bem fraca entre anos de experiência e o salário anual. Ou seja, quanto maior o tempo de experiência maior tende a ser o seu salário. Trazendo para a realidade do mercado faz total sentido pois profissionais com mais tempo de atuação acumulam habilidades, conhecimento técnico e teórico e características frequentemente associadas a cargos com maiores remunerações.
______________________________________________________________________________

Correlação salary_in_usd vs. work_year:

Correlação: 0.23

Há uma correlação bem fraca entre o ano em que o salário foi registrado e o valor do salário. Isso sugere uma tendência geral de aumento salarial ao longo do tempo, no entanto, como a correlação é fraca, isso não é suficiente para afirmar que o salário aumenta sempre de forma significativa ano após ano.
______________________________________________________________________________

Correlação years_of_experience vs. work_year:

Correlação: 0.18

Essa é a correlação mais fraca de todas, o que é de se esperar já que o ano atual de trabalho não está diretamente relacionado com o tempo de experiência do profissional, pois diferentes pessoas podem ter trajetórias e tempos de carreira distintos independentemente do ano em que estão atuando.
